# AutoML LOOCV (Paired) — FC 전체 + SHAP (5개 모델, background 43)

- **Clinical_ONLY**: `vas_t0`, `treatment_group` (2개)
- **Brain_Plus_Clinical**: 위 2개 + FC 전체 (Residualized)
- **제거 변수**: age, sex, **mean_fd** / **타겟**: vas_t1

## 목적

**어떤 ROI가 예측에 기여하는가**를 본다. 단일 모델의 결과는 그 모델의 특성일 수 있으므로,
**5개 모델에서 상위 ROI가 겹치는지**를 탐색한다.

## 대상 모델과 explainer

| 모델 | Explainer | 계산 |
|---|---|---|
| LinearRegression | `LinearExplainer` | 닫힌 식, 정확 |
| ElasticNet | `LinearExplainer` | 닫힌 식, 정확 |
| RandomForest | `TreeExplainer` | 트리 순회, 정확 |
| XGBoost | `TreeExplainer` | 트리 순회, 정확 |
| SVR_RBF | `PermutationExplainer` | model-agnostic 근사 |

`SVR_RBF`는 GPU 전용 explainer가 없으므로 CPU 기반 `PermutationExplainer`를 사용한다.
각 fold의 학습 데이터 43명 전부를 background로 사용하고 10회 permutation으로 근사한다.
이 설정은 모델 예측이나 RMSE를 바꾸지 않고, SHAP 근사의 계산량과 정밀도만 조절한다.

### training background란?

SHAP이 한 참가자의 값을 가렸을 때 대신 넣어 비교하는 **기준 데이터 분포**다.
이 노트북에서는 outer test 참가자를 절대 포함하지 않고, 해당 fold의 표준화된 학습자
43명(`X_tr`) 전부를 background로 사용한다. LinearExplainer와 SVR_RBF 모두
outer test 참가자를 제외한 43명의 경험적 학습 분포를 기준으로 삼는다.
SVR_Linear는 `coef_`가 있으므로 필요하면 별도 선형 SHAP 분석을 추가할 수 있다.

## 적용 방식 — fold 내부

LOOCV 밖에서 전체 44명으로 모델 하나를 학습해 SHAP을 뽑으면 학습에 쓴 데이터로
그 모델을 설명하는 것이 되어 과적합된 패턴을 그대로 본다. 따라서 각 fold의
`best_estimator_`에서 **그 fold의 test 샘플 1명분 SHAP**을 뽑아 44개를 모은다.

SHAP 값은 원래 **샘플마다** 나온다. 여기서는 (44, n_feat) 배열로 쌓아두고,
피처별 요약이 필요할 때 `mean(|SHAP|)`으로 압축한다. Active/Sham 비교에서는
다시 샘플 단위로 되돌려 쓴다.

## FC_LEVEL — strength 권장

`edge`(3486개)는 학습 샘플 43개당 피처 3486개다. 트리가 사실상 임의의 엣지를
고르게 되므로 SHAP이 노이즈를 설명하게 된다. `strength`(84개)를 주 분석으로 둘 것.

## 해석 시 주의

1. **SHAP은 모델이 학습한 것을 설명할 뿐, 그 학습이 옳았는지는 알려주지 않는다.**
   현재 FC 결과는 Brain 조건이 Clinical을 넘지 못한 상태다. 따라서 상위 ROI를
   "예측에 기여하는 영역"으로 단정할 수 없고, **모델 간·fold 간 일관성이 있을 때만**
   언급 가치가 있다.
2. **LinearRegression은 Brain 조건에서 p > n이라 계수가 발산한다.** morph에서
   RMSE 3.5e10이 나왔던 조건이다. 그 SHAP은 해석하지 말 것. 대조군으로만 둔다.
3. 뇌 피처는 age/sex/mean_fd로 **잔차화된 값**이다. SHAP도 "잔차화된 FC의 기여"다.
4. **SHAP으로 상위 피처를 골라 같은 데이터로 재실행하면 데이터 누수다.**
   이 분석은 보고용이며 피처 선택용이 아니다.
5. 각 환자의 SHAP은 **서로 다른 모델**에서 나왔다(fold마다 GridSearch 결과가 다름).
   일반적인 SHAP 사용법과 다른 부분이므로 논문에 명시할 것.

## morph 노트북과 달라진 점

| | morph | 이 노트북 |
|---|---|---|
| n | 45 | **44** (sub-001은 ses-t0 fmriprep 크래시로 MNI preproc BOLD 없음) |
| 교란변수 | age, sex, **eTIV** | age, sex, **mean_fd** |

## 아틀라스 표기 주의

파일명이 `FreeSurferDKT_`이지만 실제 라벨 세트는 **DK**다
(`bankssts`, `frontalpole`, `temporalpole` 포함 — DKT에서 제거된 라벨. DKT 62개, DK 68개).

```python
import numpy as np
import pandas as pd
import warnings
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
from sklearn.model_selection import LeaveOneOut, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, ElasticNet
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, r2_score
import shap

warnings.filterwarnings('ignore')
```

```python
# ==========================================================
# 0. 설정 — strength(84) / edge(3486) 선택
# ==========================================================
FC_LEVEL = 'strength'      # 'strength' 또는 'edge'

FC_FILES = {
    'strength': 'fc_strength_68_16.csv',    # ROI별 node strength 84개
    'edge':     'fc_full_84x84_edited.csv', # 상삼각 edge 3486개
}
FC_CSV = FC_FILES[FC_LEVEL]
print('사용 파일:', FC_CSV)
```

```python
# ==========================================================
# 1. DATA PREPARATION
# ==========================================================
df = pd.read_csv(FC_CSV, skipinitialspace=True)
df.columns = df.columns.str.strip()

# 앞 7개는 메타(rid, treatment_group, vas_t0, vas_t1, age, sex, mean_fd),
# 8번째 컬럼부터 끝까지가 뇌 피처
X_brain = df.iloc[:, 7:].values
brain_features = df.columns[7:].tolist()

nuisance_covariates = ['age', 'sex', 'mean_fd']    # morph의 eTIV 자리를 mean_fd가 대체
clinical_predictors = ['vas_t0', 'treatment_group']
target = 'vas_t1'

X_nuisance = df[nuisance_covariates].values
X_clinical = df[clinical_predictors].values
y = df[target].values

print(f'n = {len(df)},  brain features = {X_brain.shape[1]}  (p/n = {X_brain.shape[1]/len(df):.1f})')
```

```python
# ==========================================================
# 2. MODELS & HYPERPARAMETER SETUP
# ==========================================================
# [변경] Baseline 항목 삭제. 모든 모델이 두 데이터셋에 동일하게 적용됨.
models = {
    'LinearRegression': LinearRegression(),
    'ElasticNet': ElasticNet(max_iter=10000, random_state=42),
    'SVR_Linear': SVR(kernel='linear'),
    'SVR_RBF': SVR(kernel='rbf'),
    'RandomForest': RandomForestRegressor(random_state=42),
    'XGBoost': XGBRegressor(random_state=42, objective='reg:squarederror')
}

param_grids = {
    'LinearRegression': {},
    'ElasticNet': {
        'alpha': [0.01, 0.1, 1.0, 5.0, 10.0, 50.0, 100.0],   # 위로 확장
        'l1_ratio': [0.01, 0.05, 0.1, 0.3, 0.5, 0.7, 0.9]    # 아래로 확장
    },
    'SVR_Linear': {
        'C': [0.01, 0.1, 1.0, 10.0, 100.0, 1000.0],          # 위로
        'epsilon': [0.01, 0.1, 0.5, 1.0, 2.0]                # 위로
    },
    'SVR_RBF': {
        'C': [0.01, 0.1, 1.0, 10.0, 100.0, 1000.0],          # 위로 (최우선)
        'gamma': ['scale', 0.0001, 0.001, 0.01, 0.1],        # 아래로
        'epsilon': [0.01, 0.1, 0.5, 1.0, 2.0]
    },
    'RandomForest': {
        'n_estimators': [100, 300],
        'max_depth': [None, 2, 3, 5],
        'min_samples_leaf': [1, 3, 5]                        # split보다 leaf가 n작을 때 효과적
    },
    'XGBoost': {
        'n_estimators': [100, 300],
        'max_depth': [2, 3],                                 # n=44면 5는 과함
        'learning_rate': [0.03, 0.1],
        'reg_lambda': [1, 10]                                # 규제 추가
    },
}

DATA_TYPES = ['Clinical_ONLY', 'Brain_Plus_Clinical']
```

```python
# ==========================================================
# 3. CROSS-VALIDATION PIPELINE (LOOCV)
# ==========================================================
loo = LeaveOneOut()
# [변경] 예측을 (데이터셋, 모델) 조합별로 저장
predictions = {(dt, name): [] for dt in DATA_TYPES for name in models}
best_params_log = {(dt, name): [] for dt in DATA_TYPES for name in models}
y_true = []

# --- SHAP: fold별 test 샘플 1명분 기여도를 모은다 ---
# SVR_RBF는 CPU 기반 PermutationExplainer로 근사한다.
SHAP_BACKGROUND_SIZE = 43
SHAP_PERMUTATION_REPEATS = 10
SHAP_MODELS = {
    'LinearRegression': 'linear',
    'ElasticNet':       'linear',
    'SVR_RBF':          'permutation',
    'RandomForest':     'tree',
    'XGBoost':          'tree',
}
shap_log = {(dt, name): [] for dt in DATA_TYPES for name in SHAP_MODELS}
shap_base_log = {(dt, name): [] for dt in DATA_TYPES for name in SHAP_MODELS}
shap_additivity_error_log = {(dt, name): [] for dt in DATA_TYPES for name in SHAP_MODELS}
shap_feat = {}

print(f"Starting LOOCV: {len(models)} models x {len(DATA_TYPES)} datasets...\n")

for fold_idx, (train_idx, test_idx) in enumerate(loo.split(X_brain)):
    print(f"Processing fold {fold_idx+1}/{len(y)}...", end='\r')

    X_b_train, X_b_test = X_brain[train_idx], X_brain[test_idx]
    X_n_train, X_n_test = X_nuisance[train_idx], X_nuisance[test_idx]
    X_c_train, X_c_test = X_clinical[train_idx], X_clinical[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    # --- Nuisance Residualization ---
    nuisance_regressor = LinearRegression()
    nuisance_regressor.fit(X_n_train, X_b_train)
    X_b_train_clean = X_b_train - nuisance_regressor.predict(X_n_train)
    X_b_test_clean = X_b_test - nuisance_regressor.predict(X_n_test)

    # --- Features Stack ---
    X_train_full = np.hstack([X_b_train_clean, X_c_train])
    X_test_full = np.hstack([X_b_test_clean, X_c_test])

    y_true.append(y_test[0])

    # [변경] 두 데이터셋을 딕셔너리로 묶어 동일하게 처리
    raw_sets = {
        'Clinical_ONLY': (X_c_train, X_c_test),
        'Brain_Plus_Clinical': (X_train_full, X_test_full),
    }

    for dt, (X_tr_raw, X_te_raw) in raw_sets.items():
        scaler = StandardScaler()
        X_tr = scaler.fit_transform(X_tr_raw)
        X_te = scaler.transform(X_te_raw)

        for name, model in models.items():
            inner_cv = GridSearchCV(model, param_grids[name], cv=3,
                                    scoring='neg_mean_squared_error', n_jobs=-1)
            inner_cv.fit(X_tr, y_train)
            pred = inner_cv.best_estimator_.predict(X_te)
            predictions[(dt, name)].append(pred[0])
            best_params_log[(dt, name)].append(str(inner_cv.best_params_))

            # --- SHAP (해당 fold의 best model로 test 1명 설명) ---
            if name in SHAP_MODELS:
                est = inner_cv.best_estimator_
                kind = SHAP_MODELS[name]
                if kind == 'tree':
                    explainer = shap.TreeExplainer(est)
                    sv = np.asarray(explainer.shap_values(X_te)).reshape(-1)
                    base_value = float(np.asarray(explainer.expected_value).reshape(-1)[0])
                elif kind == 'linear':
                    # X_tr: 이 outer fold의 표준화된 학습자 43명 = SHAP 기준 분포
                    explainer = shap.LinearExplainer(est, X_tr)
                    sv = np.asarray(explainer.shap_values(X_te)).reshape(-1)
                    base_value = float(np.asarray(explainer.expected_value).reshape(-1)[0])
                elif kind == 'permutation':
                    # test 참가자를 제외한 X_tr에서만 background를 뽑는다.
                    background = shap.sample(
                        X_tr, min(SHAP_BACKGROUND_SIZE, len(X_tr)),
                        random_state=42 + fold_idx
                    )
                    explainer = shap.PermutationExplainer(
                        est.predict, background, seed=42 + fold_idx
                    )
                    max_evals = 2 * (X_tr.shape[1] + 1) * SHAP_PERMUTATION_REPEATS
                    explanation = explainer(X_te, max_evals=max_evals, silent=True)
                    sv = np.asarray(explanation.values).reshape(-1)
                    base_value = float(np.asarray(explanation.base_values).reshape(-1)[0])
                else:
                    raise ValueError(f'Unknown SHAP explainer kind: {kind}')

                # 회귀 SHAP의 local accuracy: base + sum(SHAP) == prediction
                additivity_error = abs(base_value + sv.sum() - float(pred[0]))
                shap_log[(dt, name)].append(sv)
                shap_base_log[(dt, name)].append(base_value)
                shap_additivity_error_log[(dt, name)].append(additivity_error)
                if dt not in shap_feat:
                    # X_train_full = hstack([brain, clinical]) 순서와 일치시킨다
                    shap_feat[dt] = (brain_features + clinical_predictors
                                     if dt == 'Brain_Plus_Clinical'
                                     else list(clinical_predictors))

print("\n\nLOOCV Completed.")
```

```python
# ==========================================================
# 4. EVALUATION & PAIRED COMPARISON
# ==========================================================
y_true = np.array(y_true)

results = []
for (dt, name), preds in predictions.items():
    y_pred = np.array(preds)
    corr, p_val = pearsonr(y_true, y_pred)
    results.append({
        'Data Type': dt,
        'Model': name,
        'RMSE': np.sqrt(mean_squared_error(y_true, y_pred)),
        'R2 Score': r2_score(y_true, y_pred),
        'Pearson (r)': corr,
        'p-value': p_val,
    })

results_df = pd.DataFrame(results)

# --- 기준선: 평균만 예측하는 모델 ---
baseline_rmse = np.sqrt(np.mean([(y_true[i] - np.delete(y_true, i).mean()) ** 2
                                 for i in range(len(y_true))]))
print(f"평균만 예측하는 모델의 LOOCV RMSE = {baseline_rmse:.3f}")
print("  -> 이 값보다 나쁜 모델은 예측력이 없는 것임\n")

# --- Paired 표 (행 = 모델, 열 = 데이터셋) ---
pivot = results_df.pivot(index='Model', columns='Data Type', values='RMSE')
pivot['dRMSE'] = pivot['Clinical_ONLY'] - pivot['Brain_Plus_Clinical']
pivot = pivot.sort_index()   # dRMSE 정렬 안 함 (좋은 것만 골라 읽게 되므로)

n_pos = int((pivot['dRMSE'] > 0).sum())
n_tot = len(pivot)

print("========== PAIRED COMPARISON (RMSE, +면 뇌 추가가 이득) ==========")
display(pivot)

print(f"\n부호 일관성: {n_tot}개 중 {n_pos}개가 +")
if n_pos == n_tot:
    print("  -> 전부 +. 알고리즘 선택과 무관하게 개선.")
elif n_pos == 0:
    print("  -> 전부 -. 뇌 지표가 오히려 해가 됨.")
else:
    print("  -> 부호가 섞임. 효과 없음으로 읽는 것이 타당함.")
    print("  -> 여기서 + 나온 모델만 골라 결론 내지 말 것.")

print("\n========== 전체 지표 ==========")
display(results_df.sort_values(['Model', 'Data Type']))
```

```python
# ==========================================================
# 5. VISUALIZATION (Paired)
# ==========================================================
sns.set_theme(style="ticks")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 왼쪽: 모델별 RMSE를 데이터셋으로 나눠 나란히
sns.barplot(data=results_df, x='Model', y='RMSE', hue='Data Type',
            palette=['#457B9D', '#E63946'], ax=axes[0])
axes[0].axhline(baseline_rmse, color='black', linestyle='--', linewidth=1)
axes[0].text(0.02, baseline_rmse, ' 평균만 예측', va='bottom', fontsize=9)
axes[0].set_title('RMSE: Clinical only vs. Brain added (낮을수록 좋음)')
axes[0].tick_params(axis='x', rotation=45)

# 오른쪽: dRMSE
colors = ['#2A9D8F' if v > 0 else '#E63946' for v in pivot['dRMSE']]
axes[1].bar(pivot.index, pivot['dRMSE'], color=colors)
axes[1].axhline(0, color='black', linewidth=1)
axes[1].set_title('dRMSE (+ = 뇌 지표 추가로 오차 감소)')
axes[1].set_ylabel('Clinical_ONLY - Brain_Plus_Clinical')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# ==========================================================
# 6. 하이퍼파라미터 안정성 (선택 사항)
# ==========================================================
for (dt, name), plist in best_params_log.items():
    vc = pd.Series(plist).value_counts()
    share = vc.iloc[0] / vc.sum()
    flag = '' if share >= 0.5 else '   <-- 불안정'
    print(f"[{name} / {dt}] 고유 조합 {len(vc)}개, 최빈 {share:.0%}{flag}")
```

# SHAP 분석

`share_%`는 그 피처가 전체 기여에서 차지하는 비중이다.

`mean_shap`(부호 포함)과 `mean_abs_shap`를 함께 보면 상쇄 여부를 알 수 있다 —
`mean_abs`는 큰데 `mean`이 0에 가까우면 샘플마다 기여 방향이 반대라는 뜻이다.

```python
# ==========================================================
# 7. SHAP - 모델별 피처 기여도
# ==========================================================
shap_summary = {}

for (dt, name), logs in shap_log.items():
    S = np.array(logs)                      # (n_fold, n_feat)
    feats = shap_feat[dt]
    assert S.shape[1] == len(feats), f'{dt}/{name} 피처 수 불일치: {S.shape[1]} vs {len(feats)}'

    imp = pd.DataFrame({
        'feature': feats,
        'mean_abs_shap': np.abs(S).mean(axis=0),
        'mean_shap': S.mean(axis=0),
    })
    tot = imp['mean_abs_shap'].sum()
    imp['share_%'] = imp['mean_abs_shap'] / tot * 100 if tot > 0 else 0.0
    imp = imp.sort_values('mean_abs_shap', ascending=False).reset_index(drop=True)
    shap_summary[(dt, name)] = imp

DT_MAIN = 'Brain_Plus_Clinical'

for name in SHAP_MODELS:
    imp = shap_summary[(DT_MAIN, name)]
    clin = imp[imp['feature'].isin(clinical_predictors)]['share_%'].sum()
    print(f"\n===== {name} / {DT_MAIN} =====")
    print(imp.head(10).round(4).to_string(index=False))
    print(f"  임상 2개 {clin:.1f}%  |  뇌 피처 전체 {100-clin:.1f}%")
```

```python
# ==========================================================
# 8. 모델별 상위 10개 ROI 나란히 비교
# ==========================================================
TOP_N = 10

# --- 표 1: 순위별 피처 이름 ---
side = pd.DataFrame({
    name: shap_summary[(DT_MAIN, name)]['feature'].head(TOP_N).values
    for name in SHAP_MODELS
}, index=[f'{i+1}위' for i in range(TOP_N)])

print(f"===== {DT_MAIN} | 모델별 상위 {TOP_N}개 =====")
print(side.to_string())

# --- 표 2: 몇 개 모델에서 상위 N위에 들었나 ---
from collections import Counter
cnt = Counter()
for name in SHAP_MODELS:
    cnt.update(shap_summary[(DT_MAIN, name)]['feature'].head(TOP_N).tolist())

overlap = (pd.DataFrame(cnt.items(), columns=['feature', 'n_models'])
             .sort_values('n_models', ascending=False))
for name in SHAP_MODELS:
    r = shap_summary[(DT_MAIN, name)].set_index('feature')['share_%']
    overlap[f'{name[:8]}_%'] = overlap['feature'].map(r).round(2)

print(f"\n===== 모델 간 교집합 (상위 {TOP_N} 기준, {len(SHAP_MODELS)}개 중) =====")
print(overlap.to_string(index=False))

n_all = (overlap['n_models'] == len(SHAP_MODELS)).sum()
print(f"\n  {len(SHAP_MODELS)}개 모델 전부에서 상위 {TOP_N}: {n_all}개")
print("  -> 0개에 가까우면 모델마다 다른 ROI를 골랐다는 뜻 = 노이즈일 가능성")
```

## cross-fitted local importance 빈도

각 outer fold는 서로 다른 test 참가자 1명을 설명하므로, 아래 빈도에는 모델 변화와
참가자별 local importance 차이가 함께 들어간다. 따라서 순수한 fold 안정성 검정이 아니라
**held-out 참가자 44명에서 해당 피처가 local SHAP 상위 5위에 든 빈도**로 해석한다.
빈도가 높아도 그 자체로 통계적 유의성이나 biomarker 재현성을 뜻하지 않는다.

```python
# ==========================================================
# 9. cross-fitted local importance - 상위 5위 진입 횟수
# ==========================================================
TOP_K = 5

for name in SHAP_MODELS:
    S = np.abs(np.array(shap_log[(DT_MAIN, name)]))     # (n_fold, n_feat)
    feats = np.array(shap_feat[DT_MAIN])
    n_fold = S.shape[0]

    hits = Counter()
    for row in S:
        hits.update(feats[np.argsort(-row)[:TOP_K]].tolist())

    st = (pd.DataFrame(hits.items(), columns=['feature', 'top5_count'])
            .sort_values('top5_count', ascending=False).head(10))
    st['rate_%'] = (st['top5_count'] / n_fold * 100).round(1)

    print(f"\n===== {name} / {DT_MAIN}  (n_fold={n_fold}) =====")
    print(st.to_string(index=False))
```

# 후속 실험: 고정 SHAP Top6 FC strength

FC 84개를 모두 넣었을 때는 모든 모델에서 Clinical-only보다 성능이 나빠졌다.
다만 RBF-SVR은 고차원에서 거리 계산이 노이즈에 민감하므로, background 43 SHAP에서
상위였던 **FC strength 6개를 고정 입력으로 사용하면 성능이 회복되는지** 탐색한다.

이 셀 이후에는 SHAP을 다시 계산하지 않는다. 앞에서 얻은 순위를 feature selection에만
사용하고, 기존과 같은 LOOCV + inner 3-fold GridSearch로 예측 성능만 평가한다.

비교 조건:

1. `Clinical_ONLY`: vas_t0 + treatment_group
2. `FC_All_84`: Clinical + FC strength 84개
3. `FC_SHAP_Top6`: Clinical + 고정 FC strength 6개

> 주의: Top6는 동일한 44명에서 얻은 SHAP으로 정했으므로 이 후속 결과는 탐색적이다.
> 성능이 좋아져도 독립 표본에서 재현되기 전에는 unbiased validation으로 해석하지 않는다.


```python
# ==========================================================
# 11. RBF-SVR 후속 실험 설정: SHAP Top6 FC strength
# ==========================================================
from pathlib import Path
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
from sklearn.model_selection import LeaveOneOut, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, r2_score

# 이 셀부터만 실행해도 되도록 데이터와 이전 background-43 OOF 결과를 다시 읽는다.
FC_CSV = 'fc_strength_68_16.csv'
df = pd.read_csv(FC_CSV, skipinitialspace=True)
df.columns = df.columns.str.strip()
nuisance_covariates = ['age', 'sex', 'mean_fd']
clinical_predictors = ['vas_t0', 'treatment_group']
target = 'vas_t1'
X_nuisance = df[nuisance_covariates].values
X_clinical = df[clinical_predictors].values
y = df[target].values
y_true = y.copy()
baseline_rmse = np.sqrt(np.mean([
    (y_true[i] - np.delete(y_true, i).mean()) ** 2 for i in range(len(y_true))
]))

RBF_PARAM_GRID = {
    'C': [0.01, 0.1, 1.0, 10.0, 100.0, 1000.0],
    'gamma': ['scale', 0.0001, 0.001, 0.01, 0.1],
    'epsilon': [0.01, 0.1, 0.5, 1.0, 2.0],
}

bg43_runs = sorted(
    Path('results').glob('fc_strength_loocv_shap_bg43_*'),
    key=lambda p: p.stat().st_mtime, reverse=True
)
assert bg43_runs, '먼저 background-43 SHAP 실행 결과 폴더가 필요합니다.'
BASE_BG43_RESULTS_DIR = bg43_runs[0]
prior_oof = pd.read_csv(BASE_BG43_RESULTS_DIR / 'oof_predictions.csv')
print('기존 결과:', BASE_BG43_RESULTS_DIR.resolve())

TOP6_FC_FEATURES = [
    'rh_Pallidum_fc',
    'lh_cuneus_fc',
    'rh_entorhinal_fc',
    'rh_parstriangularis_fc',
    'lh_superiorparietal_fc',
    'rh_transversetemporal_fc',
]

missing_top6 = sorted(set(TOP6_FC_FEATURES) - set(df.columns))
assert not missing_top6, f'Top6 컬럼 누락: {missing_top6}'
assert len(TOP6_FC_FEATURES) == 6 and len(set(TOP6_FC_FEATURES)) == 6

print('고정 Top6 FC features:')
for rank, feature in enumerate(TOP6_FC_FEATURES, 1):
    print(f'{rank}. {feature}')
print('최종 입력 수: clinical 2 + FC 6 = 8')
```

```python
# ==========================================================
# 12. 고정 Top6 FC + Clinical: RBF-SVR LOOCV
# ==========================================================
X_top6 = df[TOP6_FC_FEATURES].values
top6_predictions = []
top6_best_params = []
top6_y_true = []

for fold_idx, (train_idx, test_idx) in enumerate(LeaveOneOut().split(X_top6)):
    print(f'Processing Top6 fold {fold_idx + 1}/{len(df)}...', end='\r')

    X_f_train, X_f_test = X_top6[train_idx], X_top6[test_idx]
    X_n_train, X_n_test = X_nuisance[train_idx], X_nuisance[test_idx]
    X_c_train, X_c_test = X_clinical[train_idx], X_clinical[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    # 기존 FC-all 실험과 동일: outer-training 43명에서 nuisance residualization
    nuisance_regressor = LinearRegression()
    nuisance_regressor.fit(X_n_train, X_f_train)
    X_f_train_clean = X_f_train - nuisance_regressor.predict(X_n_train)
    X_f_test_clean = X_f_test - nuisance_regressor.predict(X_n_test)

    X_train_top6 = np.hstack([X_f_train_clean, X_c_train])
    X_test_top6 = np.hstack([X_f_test_clean, X_c_test])

    scaler = StandardScaler()
    X_train_top6 = scaler.fit_transform(X_train_top6)
    X_test_top6 = scaler.transform(X_test_top6)

    search = GridSearchCV(
        SVR(kernel='rbf'),
        RBF_PARAM_GRID,
        cv=3,
        scoring='neg_mean_squared_error',
        n_jobs=-1,
    )
    search.fit(X_train_top6, y_train)
    pred = float(search.best_estimator_.predict(X_test_top6)[0])

    top6_predictions.append(pred)
    top6_best_params.append(search.best_params_)
    top6_y_true.append(float(y_test[0]))

print('\nTop6 LOOCV Completed.')
top6_predictions = np.asarray(top6_predictions)
top6_y_true = np.asarray(top6_y_true)
assert len(top6_predictions) == len(df) == 44
assert np.all(np.isfinite(top6_predictions))
assert np.allclose(top6_y_true, y_true)
```

```python
# ==========================================================
# 13. Clinical vs FC-all vs 고정 Top6 비교 및 저장
# ==========================================================
def load_prior_rbf_predictions(data_type):
    rows = (prior_oof[
        (prior_oof['Data Type'] == data_type) & (prior_oof['Model'] == 'SVR_RBF')
    ].sort_values('fold'))
    assert len(rows) == len(df), f'{data_type} OOF 행 수 오류: {len(rows)}'
    assert rows['rid'].tolist() == df['rid'].tolist(), f'{data_type} rid 순서 불일치'
    assert np.allclose(rows['y_true'].values, y_true), f'{data_type} target 불일치'
    return rows['y_pred'].to_numpy()

rbf_comparison_predictions = {
    'Clinical_ONLY': load_prior_rbf_predictions('Clinical_ONLY'),
    'FC_All_84': load_prior_rbf_predictions('Brain_Plus_Clinical'),
    'FC_SHAP_Top6': top6_predictions,
}
n_features_by_condition = {
    'Clinical_ONLY': 2,
    'FC_All_84': 86,
    'FC_SHAP_Top6': 8,
}

comparison_rows = []
clinical_sqerr = (y_true - rbf_comparison_predictions['Clinical_ONLY']) ** 2
for condition, pred_values in rbf_comparison_predictions.items():
    r, p_value = pearsonr(y_true, pred_values)
    sqerr = (y_true - pred_values) ** 2
    comparison_rows.append({
        'Condition': condition,
        'N': len(y_true),
        'n_features': n_features_by_condition[condition],
        'RMSE': np.sqrt(mean_squared_error(y_true, pred_values)),
        'R2 Score': r2_score(y_true, pred_values),
        'Pearson (r)': r,
        'p-value': p_value,
        'dRMSE_vs_Clinical': (
            np.sqrt(mean_squared_error(y_true, rbf_comparison_predictions['Clinical_ONLY']))
            - np.sqrt(mean_squared_error(y_true, pred_values))
        ),
        'participants_with_lower_SE_than_Clinical': (
            int(np.sum(sqerr < clinical_sqerr)) if condition != 'Clinical_ONLY' else np.nan
        ),
    })

rbf_top6_metrics = pd.DataFrame(comparison_rows).sort_values('RMSE').reset_index(drop=True)
print('===== RBF-SVR INPUT COMPARISON =====')
display(rbf_top6_metrics)

top6_gain = float(
    rbf_top6_metrics.loc[rbf_top6_metrics['Condition'] == 'FC_SHAP_Top6', 'dRMSE_vs_Clinical'].iloc[0]
)
if top6_gain > 0:
    print(f'Top6가 Clinical-only보다 RMSE {top6_gain:.4f} 개선되었습니다.')
else:
    print(f'Top6가 Clinical-only보다 RMSE {-top6_gain:.4f} 악화되었습니다.')
print('주의: 동일 cohort의 SHAP으로 Top6를 정한 탐색적 후속 결과입니다.')

fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(data=rbf_top6_metrics, x='Condition', y='RMSE', color='#457B9D', ax=ax)
ax.axhline(baseline_rmse, color='black', linestyle='--', linewidth=1, label='LOOCV mean baseline')
ax.set_title('RBF-SVR: Clinical vs FC all vs fixed SHAP Top6')
ax.set_ylabel('OOF RMSE (lower is better)')
ax.tick_params(axis='x', rotation=15)
ax.legend()
fig.tight_layout()

# 기존 background-43 결과 폴더에 후속 결과를 이어서 저장
TOP6_RESULTS_DIR = BASE_BG43_RESULTS_DIR
rbf_top6_metrics.to_csv(TOP6_RESULTS_DIR / 'rbf_svr_top6_metrics.csv', index=False)
pd.DataFrame({
    'fold': np.arange(1, len(df) + 1),
    'rid': df['rid'].values,
    'treatment_group': df['treatment_group'].values,
    'y_true': y_true,
    'pred_Clinical_ONLY': rbf_comparison_predictions['Clinical_ONLY'],
    'pred_FC_All_84': rbf_comparison_predictions['FC_All_84'],
    'pred_FC_SHAP_Top6': rbf_comparison_predictions['FC_SHAP_Top6'],
}).to_csv(TOP6_RESULTS_DIR / 'rbf_svr_top6_oof_predictions.csv', index=False)
pd.DataFrame({
    'fold': np.arange(1, len(df) + 1),
    'rid': df['rid'].values,
    'best_params': [json.dumps(x, sort_keys=True) for x in top6_best_params],
}).to_csv(TOP6_RESULTS_DIR / 'rbf_svr_top6_best_params.csv', index=False)
fig.savefig(TOP6_RESULTS_DIR / 'rbf_svr_top6_rmse.png', dpi=200, bbox_inches='tight')
plt.show()
print('Saved Top6 follow-up:', TOP6_RESULTS_DIR.resolve())
```

FC 전체를 넣었을 때보다 Top6만 넣었을 때 성능이 크게 회복됐습니다.
- Clinical-only RBF보다 RMSE 0.2150, 약 9.7% 개선
